# PHÂN TÍCH MÔ TẢ
# Mục Tiêu:
#    - Tóm tắt dl theo nhóm, theo tgian
#    - Tạo bảng tổg hợp (pivot table)
#    - Trực quan hóa kết quả bằng biểu đồ để thấy xu hướng

```python
# tổng doanh thu và sl bán theo từng sản phẩm
df = pd.read_csv('sales_data.csv')
prod_summary = df.groupby("Product").agg(
    Total_Quanti = ("Quantity", "sum"),
    Total_Revenue = ("Revenue", "sum"),
    Avg_Quanti = ("Quantity", "mean")
).reset_index()

print(prod_summary)
print('\n')

df["Date"] = pd.to_datetime(df["Date"]) 
print(df.dtypes)
print('\n')

df['Revenue'] = df['Revenue'].fillna(df['Revenue'].mean())
print('Revenue: ', df['Revenue'])
print('\n')

df['Month'] = df['Date'].dt.to_period('M')
print(df[["Date", "Month"]].head())
print('\n')

monthly_rev = df.groupby('Month')['Revenue'].sum()
print(monthly_rev)
print('\n')

pivot_table = df.pivot_table(
    values = "Revenue",
    index = 'Month',
    columns="Product",
    aggfunc="sum"
)

print(pivot_table)
print('\n')

# vẽ biểu đồ so sánh doanh thu theo sp
prod_summary.plot(
    x="Product",
    y="Total_Revenue",
    kind = "bar",
    legend = False,
    color = 'skyblue'
)
plt.title("tổng doanh thu theo sản phẩm")
plt.ylabel("Doanh thu")
plt.show()
print('\n')

# vẽ biểu đồ doanh thu theo tháng
monthly_rev.plot(kind="line", marker="o")
plt.title("Doanh thu theo tháng")
plt.show()
print('\n')

# trực quan hóa tương quan bằng heatmap
sns.heatmap(pivot_table, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Doanh thu theo tháng và sản phẩm")
plt.ylabel("Tháng")
plt.xlabel("Sản phẩm")
plt.show()
print('\n')
```

# Thực Hành Phân Tích Chẩn Đoán

```python
# ma trận tương quan
# kiểm tra tương quan giữa các biến só
print('KIỂM TRA TƯƠNG QUAN GIỮA CÁC BIẾN SỐ')
corr_matrix = df[["Quantity", "Revenue"]].corr()
print(corr_matrix)
print('-----------------------------------------------')
print('\n')

# trực quan hóa tương quan bằng heatmap
print('TRỰC QUAN HÓA TƯƠNG QUAN = HEATMAP')
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", center=0)
plt.title("Ma trận tương quan giữa Quantity và Revenue")
plt.show()
print('------------------------------------------------')
print('\n')

# phân tích mối quan hệ Quantity - Revenue bằng scatter
print('PHÂN TÍCH MQH QUANTITY - REVENUE = SCATTER')
plt.scatter(df["Quantity"], df["Revenue"], alpha=0.5)
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.title("Mối quan hệ giữa Quantity và Revenue")
plt.show()
print('------------------------------------------------')
print('\n')

# xây dựng mô hình hồi quy tuyến tính đơn giản
print('XÂY DỰNG MÔ HÌNH HỒI QUY TUYẾN TÍNH ĐƠN GIẢN')
from sklearn.linear_model import LinearRegression
import numpy as np
X = df[['Quantity']] # biến độc lập
y = df['Revenue'] # biến phụ thuộc

model = LinearRegression()
model.fit(X, y)

print("Hệ số góc (slope):", model.coef_[0])
print("Hệ số chặn (intercept):", model.intercept_)
print("R^2 score:", model.score(X, y))
print('Slope: cho biết trung bình doanh thu tăng bnhiu khi bán thêm 1 đơn vị sản phẩm')
print('Intercept: dthu dự đoán khi số lượng = 0 (thường gần 0)')
print('R^2 score: mức độ giải thích biến thiên doanh thu bởi số lượng')
print('-------------------------------------------')
print('\n')

print('VẼ ĐƯỜNG HỒI QUY TRÊN SCATER PLOT')
y_red = model.predict(X)
plt.scatter(df["Quantity"], df["Revenue"], alpha = 0.5, label="Dữ liệu thực tế")
plt.plot(df["Quantity"], y_red, color="red", linewidth=2, label="Đường Hồi Quy")
plt.xlabel('Quantity')
plt.ylabel('Revenue')
plt.title('Hồi quy tuyến tính: Quantity vs Revenue')
plt.legend()
plt.show()
print('-------------------------------------------')
print('\n')

# tổng doanh thu theo tháng
print('CHUẨN BỊ DỮ LIỆU THEO THÁNG')
monthly_df = df.groupby('Month')['Revenue'].sum().reset_index()

# đổi month về dạng datetime (giá trị giữa tháng)
print('CHUYỂN MONTH DẠNG DATETIME (.dt.to_timestamp())')
monthly_df['Month'] = monthly_df['Month'].dt.to_timestamp()
print(monthly_df)
print('--------------------------------------------')
print('\n')

# chuyển thời gian thành số thứ tự (0, 1, 2, . . .)
print('CHUYỂN TGIAN THÀNH SỐ THỨ TỰ')
monthly_df["t"] = range(len(monthly_df))
print(monthly_df['t'])
print('----------------------------------------------')
print('\n')

#chia dl Train/test
print('CHIA DỮ LIỆU TRAIN / TEST')
from sklearn.model_selection import train_test_split
X = monthly_df[["t"]]
y = monthly_df['Revenue']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, shuffle=False)
print('\n')
print('----------------------------------------------')
print('\n')

# huấn luyện mô hình hồi quy tuyến tính
from sklearn.linear_model import LinearRegression
print('HUẤN LUYỆN MODULE HỒI QUY TUYẾN TÍNH')
model = LinearRegression()
model.fit(X_train, y_train)
print("Hệ số góc:", model.coef_[0])
print("Hệ số chặn:", model.intercept_)
print('----------------------------------------------')
print('\n')

import matplotlib.pyplot as plt
print('DỰ BÁO VÀ ĐÁNH GIÁ')
y_red = model.predict(X_test)
plt.plot(monthly_df['Month'], monthly_df['Revenue'],label="Doanh Thu Thực Tế", marker="o")
plt.plot(X_test.index, y_red, label="Dự Báo", marker='x',linestyle='--')
plt.title("Dự báo doanh thu theo tháng")
plt.legend()
plt.show()

from sklearn.metrics import mean_squared_error, r2_score
print('MSE:', mean_squared_error(y_test, y_red))
print('R^2:', r2_score(y_test, y_red))
print('MSE: càng nhỏ càng tốt')
print('R^2: càg gần 1 càng tốt (mô hình giải thích tốt dữ liệu)')
print('----------------------------------------------')
print('\n')

# dự báo thág tiếp theo
print('DỰ BÁO THÁNG TIẾP THEO')
next_t = np.array([[monthly_df["t"].max() + 1]])
next_month_prediction = model.predict(next_t)
print('Dự báo doanh thu tháng tiếp theo', next_month_prediction[0])
print('--------------------------------------------------------------------')
print('\n')

# tính doanh thu trung bình mỗi đơn vị bán
print('TÍNH DOANH THU TRUNG BÌNH MỖI ĐƠN VỊ BÁN')
prod_summary['Revenue_per_unit'] =  (prod_summary['Total_Revenue'] /  prod_summary['Total_Quanti'])
print(prod_summary[['Product', 'Revenue_per_unit']])
print('--------------------------------------------------------------------')
print('\n')

# thiết lập mô hình tối ưu hóa
# dùng pulp để giải bài toán phân bổ ngân sách
import pulp

# biến quyết định: ngsach đc phân bố đều cho mọi Sản phẩm (0 - 100)
print('NGÂN SÁCH CHIA ĐỀU CHO MỌI SẢN PHẨM')
prods = prod_summary['Product'].to_list()
budget = 100
print(prods)
print('--------------------------------------------------------------------')
print('\n')

# KHỞI TẠO BÀI TOÁN TỐI ƯU
# pulp.LpProblem('Ten_Bai_Toan', muctieu đề ra (pupl.LpMaximize))
print('KHỞI TẠO BÀI TOÁN TỐI ƯU')
prob = pulp.LpProblem('Marketing_Optimization', pulp.LpMaximize)
print(prob)
print('\n')
print('--------------------------------------------------------------------')
print('\n')

# BIẾN QUYẾT ĐỊNH
print('BIẾN QUYẾT ĐỊNH')
alloc = pulp.LpVariable.dicts('Budget', prods, lowBound = 0)
print(alloc)
print('--------------------------------------------------------------------')
print('\n')

# HÀM MỤC TIÊU: TỐI ĐA HÓA DOANH THU DỰ KIẾN
print('TỐI ĐA HÓA DOANH THU DỰ KIẾN')
prob += pulp.lpSum([
    alloc[p] * prod_summary.loc[
    prod_summary['Product'] == p,
    'Revenue_per_unit'
    ].values[0]
    for p in prods
])
print(prods)
print('--------------------------------------------------------------------')
print('\n')

# GIẢI BÀI TOÁN
prob.solve()
for p in prods:
        print(f'Phân bố cho {p}: {alloc[p].value()}')
```

```python
# tạo chatbot đơn giản (rule-based)

print('Chào bạn ! Tôi là Chatbot đơn giản .')
while True:
    user_input = input("Bạn: ").lower()
    if "chào" in user_input:
        print("ChatBot: Xin chào ! Rất vui được gặp bạn .")
    elif "tên" in user_input:
        print("ChatBot: Tôi là ChatBot rule-based, chưa thông minh lắm .")
    elif "tạm biệt" in user_input:
        print("ChatBot: Hẹn gặp lại !")
        break
    else:
        print("ChatBot: Xin lỗi, tôi chưa hiểu ý bạn .")
```

```python
from transformers import pipeline

# tải mô hình hội thoại (transformers model)
chatbot = pipeline("text-generation", model="gpt2")

while True:
    user_input = input("Bạn: ")
    if user_input.lower() in ["thoát", "tạm biệt"]:
        print("ChatBot: Hẹn gặp lại !")
        break

    response = chatbot(user_input, max_length=50, do_sample=True, temperature=0.7)
    print("ChatBot:", response[0]['generation_text'])
```

## MỤC TIÊU
 * LÀM QUEN VỚI QUY TRÌNH XÂY DỰNG MÔ HÌNH TRONG Scikit-learnn
 * THẤY ĐƯỢC CÁCH MÔ HÌNH HỌC TỪ DỮ LIỆU VÀ DỰ ĐOÁN KẾT QUẢ MỚI

```python
# Bước 1: chuẩn bị dữ liệu mẫu
import pandas as pd

# tạo mẫu dataset (chiều cao, cân nặng, giới tính)
data = {
    'chieu_cao': [150, 160, 165, 170, 175, 180, 185],
    'can_nang': [45, 55, 60, 65, 70, 80, 85],
    'gioi_tinh': ['Nữ', 'Nữ', 'Nữ', 'Nam', 'Nam', 'Nam', 'Nam']
}

df = pd.DataFrame(data)
print('Chuẩn bị data mẫu: ')
print(df)
print('\n')
print('================================================================================================')
print('\n')

# Bước 2: tiền xử lí dữ liệu
# chuyển biến phân loại giới tính từ Nam = 1, Nữ = 0
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['gioi_tinh'] = le.fit_transform(df['gioi_tinh'])
print('Chuyển giới tính thành nhãn số: ')
print(df)
print('\n')
print('================================================================================================')
print('\n')

# Bước 3: chia dữ liệu train/test
from sklearn.model_selection import train_test_split
X = df[['chieu_cao', 'can_nang']]
y = df['gioi_tinh']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Bước 4: huấn luyện mô hình kNN
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# Bước 5: Dự đoán và đánh giá
y_pred = model.predict(X_test)
from sklearn.metrics import accuracy_score

print("Dự đoán:", y_pred)
print("Thực tế:", y_test.values)
print("Độ chính xác:", accuracy_score(y_test, y_pred))
print('\n')
print('================================================================================================')
print('\n')

# Bước 6: thử dự đoán cho người mới
new_data = [[172, 68]]
result = model.predict(new_data)
print("Dự đoán giới tính cho (172cm, 68kg):", le.inverse_transform(result)[0])
print('\n')
print('================================================================================================')
print('\n')
```

## THỰC HÀNH NÂNG CAO: PHÂN LOẠI HOA IRIS VỚI SCIKIT-LEARN
* Làm quen với dataset kinh điển Iris (150 mẫu, 3 loại hoa)
* Thực hành toàn bộ quy trình ML: train/test slpit, huấn luyện, đánh giá, trực quan hóa

```python
from sklearn.datasets import load_iris
import pandas as pd

# Bước 1: nạp dữ liệu
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

print(df.head())
print(df['species'].value_counts())
# bộ dữ liệu có sẵn trong thư viện scikit-learn (sklearn), chứ không phải đọc từ file CSV hay Excel trên máy tính.
print('\n')
print('================================================================================================')
print('\n')

# Bước 2: chia train/test
from sklearn.model_selection import train_test_split
X = df[iris.feature_names] # 4 đặc trưng: sepal & petal
y = df['species'] # nhãn (0, 1, 2)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Bước 3: huấn luyện mô hình kNN
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
print('\n')

# Bước 4: đánh giá mô hình
from sklearn.metrics import accuracy_score, classification_report
y_pred = model.predict(X_test)
print("Độ chính xác:", accuracy_score(y_test, y_pred))
print("Báo cáo phân loại:\n", classification_report(y_test, y_pred, target_names=iris.target_names)) 
print('\n')
print('================================================================================================')
print('\n')

# Bước 5: trực quan hóa kết quả
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,5))
sns.scatterplot(x='petal length (cm)', y='petal width (cm)', data=df, hue='species', palette='Set1')
plt.title('Phân bố loài hoa theo kích thước cánh hoa') 
plt.show() 

# Bước 6: thử dự đoán mới
sample = [[5.0, 3.4, 1.5, 0.2]]  # Một mẫu mới 
prediction = model.predict(sample) 
print("Dự đoán loài hoa:", iris.target_names[prediction][0]) 
```

# MỤC TIÊU
* Làm quen với thư viện OpenCV
* Hiểu cách xử lý ảnh số: đọc ảnh, chuyển đổi màu, phát hiện cạnh
* Quan sát sự khác biệt giữa ảnh gốc và ảnh đã xử lí

```python
# Bước 1: đọc và hiển thị ảnh
import cv2
import matplotlib.pyplot as plt

# đọc ảnh (dùng ảnh có sẵn (internet))
img = cv2.imread('image.png') # phải nằm trong thư mục làm việc

# chuyển từ BGR (OpenCV) sang RGB (Matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title('Ảnh gốc')
plt.axis('off')
plt.show()

# Bước 2: chuyển ảnh sang thang xám
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title('Ảnh xám')
plt.axis('off')
plt.show()

# Bước 3: phát hiện cạnh bằng Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)
plt.imshow(edges, cmap='gray')
plt.title('Phát hiện cạnh (Canny)')
plt.axis('off')
plt.show()

# Bước 4: làm mờ ảnh (Gaussian Blur)
blurred = cv2.GaussianBlur(img_rgb, (15, 15), 0)
plt.imshow(blurred)
plt.title('Ảnh sau khi làm mờ')
plt.axis('off')
plt.show()

# thử đọc ảnh từ webcam
cap = cv2.VideoCapture(0) 
ret, frame = cap.read() 
cap.release() 
plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)) 
plt.title('Ảnh chụp từ webcam') 
plt.axis('off') 
plt.show()
```

# MỤC TIÊU
* Làm quen với transformer models (BERT, DistilBERT, GPT, v.v.) trong xử lí ngôn ngữ tự nhiên
* Trải nghiệm pipeline của Hugging Face để phân tích cảm xúc (sentiment analysis) hoặc dịch văn bản
* Hiểu cách ứng dụng mô hình có sẵn (pre-trained models) vào bài toán thực tế mà không cần huấn luyện

```python
# tạo pipeline phân tích cảm xúc
from transformers import pipeline
classifier = pipeline("sentiment-analysis")

# văn bản mẫu 
texts = ["I love this products ! It's really amazing.", "This is the worst service I've ever experienced."]

results = classifier(texts)
for text, result in zip(texts, results):
    print(f'Text: {text}\nSentiment: {result['label']}, Score: {result['score']:.4f}\n')

# thử nghiệm dịch máy
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-en-vi")
text = "Machine learning is transforming the world of technology."
translation = translator(text, max_length=100)
print("Original:", text)
print("Translated:", translation[0]['translation_text'])
```

# A - Phân tích cảm xúc đánh giá sản phẩm
# Mục tiêu
* Kết hợp kiến thức EDA (Chương 4) và NLP (Chương 6)
* Áp dụng Hugging Face pipeline để phân tích cảm xúc review sản phẩm
* Trực quan hóa kết quả bằng biểu đồ để rút ra insight

```python
# Bước 1: tạo data
import pandas as pd
data = {
    "reivew": [
        "I love this products ! It's really amazing",
        "The quanlity is very bad, totally disappointed",
        "Good value for money, will buy again",
        "Terrible customer service, never coming back !",
        "Excellent packaging and fast delivery."
    ]
}

df = pd.DataFrame(data) 
df.to_csv("reviews.csv", index=False) 
print(df) 

# Bước 2: phân tích cảm xúc với Hugging Face
from transformers import pipeline
classifier = pipeline("sentiment-analysis")

 
df['sentiment'] = df['review'].apply(lambda x: 
classifier(x)[0]['label']) 
df['score'] = df['review'].apply(lambda x: 
classifier(x)[0]['score']) 
print(df)

import seaborn as sns 
import matplotlib.pyplot as plt 
sns.countplot(x='sentiment', data=df, palette='Set2') 
plt.title("Phân bố cảm xúc review sản phẩm") 
plt.xlabel("Loại cảm xúc") 
plt.ylabel("Số lượng") 
plt.show() 
```
